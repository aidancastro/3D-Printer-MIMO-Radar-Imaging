```python
def plot_25_yz(vol_stack, Y=None, Z=None, use_mip=True):
    """
    vol_stack: (Ny, Nx, Nz, N) complex. Shows up to 25 scans.
    Y, Z: optional 1D axes for labeling. If None, index units are used.
    use_mip=True -> YZ max-intensity projection; False -> YZ mid-X slice.
    """
    vols = np.asarray(vol_stack)
    Ny, Nx, Nz, N = vols.shape
    N = min(N, 25)

    # global 0 dB normalization across all scans
    mag = np.abs(vols[..., :N]).astype(np.float64)
    ref = mag.max() + 1e-12
    vols_db = 20*np.log10(mag/ref + 1e-12)

    # consistent display range
    vmin, vmax = np.percentile(vols_db, [50, 99])

    yi = Y if Y is not None else np.arange(Ny)
    zi = Z if Z is not None else np.arange(Nz)
    xm = Nx // 2

    fig, axes = plt.subplots(5, 5, figsize=(15, 15), constrained_layout=True)
    for i, ax in enumerate(axes.flat):
        if i < N:
            yz = vols_db[:, xm, :, i] if not use_mip else vols_db[..., i].max(axis=1)
            ax.imshow(yz, origin="lower", aspect="auto",
                      extent=[zi.min(), zi.max(), yi.min(), yi.max()],
                      vmin=vmin, vmax=vmax)
            ax.set_title(f"{i+1}")
        ax.set_xticks([]); ax.set_yticks([])
    plt.show()
```

```python
vol_stack = np.stack(volumes, axis=3)
plot_25_yz(vol_stack, use_mip=True)
```
