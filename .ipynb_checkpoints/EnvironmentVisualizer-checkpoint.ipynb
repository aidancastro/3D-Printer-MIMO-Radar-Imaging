```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

%matplotlib widget

# ---- Original VK-74 list: [Y, X, Z] ----
Vant_yxz = np.array([
  [-0.0275,-0.0267,0],[-0.0253,-0.0267,0],[-0.0231,-0.0267,0],[-0.0209,-0.0267,0],
  [-0.0187,-0.0267,0],[-0.0165,-0.0267,0],[-0.0143,-0.0267,0],[-0.0122,-0.0267,0],
  [-0.01,-0.0267,0],[-0.0078,-0.0267,0],[-0.0056,-0.0267,0],[-0.0034,-0.0267,0],
  [-0.0012,-0.0267,0],[0.0009,-0.0267,0],[0.0031,-0.0267,0],[0.0053,-0.0267,0],
  [0.0075,-0.0267,0],[0.0097,-0.0267,0],[0.0119,-0.0267,0],[0.0141,-0.0267,0],
  [0.0274,-0.0133,0],[0.0274,-0.0112,0],[0.0274,-0.0091,0],[0.0274,-0.007,0],
  [0.0274,-0.0049,0],[0.0274,-0.0028,0],[0.0274,-0.0007,0],[0.0275,0.0014,0],
  [0.0275,0.0035,0],[0.0275,0.0056,0],[0.0275,0.0078,0],[0.0275,0.0099,0],
  [0.0275,0.012,0],[0.0274,0.0141,0],[0.0274,0.0162,0],[0.0275,0.0183,0],
  [0.0275,0.0204,0],[0.0275,0.0225,0],[0.0275,0.0246,0],[0.0275,0.0267,0]
], dtype=np.float32)

# Convert to (x=vertical, y=horizontal, z=downrange)
Vant_xyz = Vant_yxz[:, [1, 0, 2]]

# Split by hardware: RX first 20, TX last 20
RX_hw = Vant_xyz[:20]
TX_hw = Vant_xyz[20:]

# =========================
# TX & RX ANTENNA POSITIONS
# =========================
gap = 0.0133
margin = 0.003  # keep small separation from zero axes

# --- TX x-spacing preserved ---
tx_x_raw  = TX_hw[:, 0]
tx_order  = np.argsort(tx_x_raw)[::-1]            # TX1(top) → TX20(bottom)
tx_sorted = tx_x_raw[tx_order]
tx_span   = tx_sorted.max() - tx_sorted.min()

# --- RX y-spacing preserved ---
rx_y_raw  = RX_hw[:, 1]
rx_order  = np.argsort(rx_y_raw)                  # RX21(left) → RX40(right)
rx_y_sorted = rx_y_raw[rx_order]

# Set RX x-position so TX20.x = RX21.x + gap
RX_x_val = 0.0  # start RX at zero for now
TX_x = (tx_sorted - tx_sorted.min()) + (RX_x_val + gap)

# Set RX y base so TX_y = RX21_y - gap
RX_y = rx_y_sorted - rx_y_sorted.min()  # preserve spacing from RX21
TX_y_val = RX_y.min() - gap

# If any coordinates are negative, shift everything up/right together
shift_x = 0
shift_y = 0

if TX_x.min() < 0 or RX_x_val < 0:
    shift_x = abs(min(TX_x.min(), RX_x_val)) + margin
if TX_y_val < 0 or RX_y.min() < 0:
    shift_y = abs(min(TX_y_val, RX_y.min())) + margin

TX_x += shift_x
RX_x_val += shift_x
RX_y += shift_y
TX_y_val += shift_y

# Build RX positions
RX_x = np.full_like(RX_y, RX_x_val)
TX_z = np.zeros_like(TX_x)
RX_z = np.zeros_like(RX_y)

TX_plot = np.column_stack([TX_x, np.full_like(TX_x, TX_y_val), TX_z])
RX_plot = np.column_stack([RX_x, RX_y, RX_z])
tx_labels = [f"{i}" for i in range(1, 21)]
rx_labels = [f"{i}" for i in range(21, 41)]

# -----------------
# Plotting antennas
# -----------------
def to_plot(arr_xyz):
    return np.column_stack([arr_xyz[:,1], arr_xyz[:,0], arr_xyz[:,2]])

TXp = to_plot(TX_plot)
RXp = to_plot(RX_plot)

fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(projection="3d")
ax.set_box_aspect((1, 1, 0.05))

ax.scatter(TXp[:,0], TXp[:,1], TXp[:,2], s=110, c="red",  edgecolors="black", label="TX 1–20")
ax.scatter(RXp[:,0], RXp[:,1], RXp[:,2], s=110, c="blue", edgecolors="black", label="RX 21–40")

stroke = [pe.withStroke(linewidth=3.5, foreground="white")]
for (xp, yp, zp), lab in zip(TXp, tx_labels):
    ax.text(xp, yp, zp, lab, fontsize=12, color="red", path_effects=stroke, ha='center', va='center')
for (xp, yp, zp), lab in zip(RXp, rx_labels):
    ax.text(xp, yp, zp, lab, fontsize=12, color="blue", path_effects=stroke, ha='center', va='center')

ax.set_xlabel("y (horizontal)")
ax.set_ylabel("x (vertical)")
ax.set_zlabel("z (downrange)")
ax.view_init(elev=12, azim=25)
ax.legend(loc="upper left")
plt.tight_layout()
plt.show()

# ------------------------
# Print antenna coordinates
# ------------------------
def print_antenna_coordinates():
    print("\n=== TX Antennas (1–20) ===")
    for label, coords in zip(tx_labels, TX_plot):
        x, y, z = coords
        print(f"TX{label:>2}: x={x:.4f} m, y={y:.4f} m, z={z:.4f} m")

    print("\n=== RX Antennas (21–40) ===")
    for label, coords in zip(rx_labels, RX_plot):
        x, y, z = coords
        print(f"RX{label:>2}: x={x:.4f} m, y={y:.4f} m, z={z:.4f} m")

print_antenna_coordinates()

# ------------------------
# Sanity check constraints
# ------------------------
tx20_x = TX_x.min()
rx21_x = RX_x_val
rx21_y = RX_y.min()
tx_y   = TX_y_val

print(f"\nCheck X-gap: TX20.x - RX21.x = {tx20_x - rx21_x:.4f} m (target {gap})")
print(f"Check Y-gap: RX21.y - TX.y   = {rx21_y - tx_y:.4f} m (target {gap})")
print(f"Min X = {min(TX_x.min(), RX_x_val):.4f}, Min Y = {min(TX_y_val, RX_y.min()):.4f}")
```
